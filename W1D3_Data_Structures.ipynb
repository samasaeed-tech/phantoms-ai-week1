```python
students = {
    'Sama': 90,
    'Ramy': 'Absent',
    'Sara': 'Absent',
    'Mohamed': 70
}

def calculate_average_score(input_data):
    collected_scores = []

    for name, mark in input_data.items():
        try:
            numeric_value = float(mark)
            collected_scores.append(numeric_value)
        except (ValueError, TypeError):
            continue

    try:
        avg = sum(collected_scores) / len(collected_scores)
        return avg
    except ZeroDivisionError:
        return 0.0

final_score = calculate_average_score(students)
print("Average score:", final_score)
```
