# Python Debugging Challenge

بعد مراجعة الكود، اكتشفت 3 أخطاء منطقية (Logical Bugs) بتأثر على صحة النتائج:

1. **تجاهل آخر عنصر في القائمة:**
   * **السبب:** استخدام `range(len(scores_list) - 1)` بيخلي اللوب تقف قبل آخر عنصر، وبالتالي آخر درجة مش بتتحسب معانا في المجموع ولا في النجاح.
   * **التصحيح:** نعمل لفة مباشرة على العناصر `for score in scores_list:`.

2. **تجاهل الدرجة 50 من قائمة الناجحين:**
   * **السبب:** الشرط مكتوب `score > 50` بينما المطلوب إن درجة 50 تعتبر درجة نجاح (>= 50).
   * **التصحيح:** نعدل الشرط يكتب `score >= 50`.

3. **احتمالية انهيار البرنامج لو القائمة فاضية:**
   * **السبب:** حساب المتوسط بيقسم على طول `len(scores_list)`، ولو القائمة فاضية القسمة على صفر هتعمل `ZeroDivisionError` والبرنامج هيقف.
   * **التصحيح:** استخدام `try-except` للتعامل مع الحالات الغلط أو القوائم الفاضية بأمان.

```python
def process_team_scores(scores_list):

    try:
        # التأكد إن القائمة مش فاضية
        if not scores_list:
            print("القائمة فارغة، لا يمكن حساب المتوسط.")
            return 0.0, []

        total_score = 0
        passed_members = []

        for score in scores_list:
            if score >= 50:
                passed_members.append(score)
            total_score += score

        average = total_score / len(scores_list)
        return average, passed_members

    except (TypeError, ValueError) as e:
        print(f"حدث خطأ في أنواع البيانات: {e}")
        return 0.0, []

# تجربة الكود (Test Cases)

# 1. التجربة العادية
sample_scores = [45, 50, 75, 90, 60]
avg, passed = process_team_scores(sample_scores)
print("Normal Case -> Average:", avg, "| Passed:", passed)

# 2. تجربة قائمة فارغة
avg_empty, passed_empty = process_team_scores([])
print("Empty Case -> Average:", avg_empty, "| Passed:", passed_empty)

# 3. تجربة بيانات غير رقمية
avg_err, passed_err = process_team_scores([80, 'Absent', 60])
print("Error Case -> Average:", avg_err, "| Passed:", passed_err)
```
